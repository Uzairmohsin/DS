```python
import matplotlib.pyplot as plt
years = [2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020]

kohli = [845, 1025, 980, 1120, 1185, 1350, 1280, 1425, 1310, 1195]

rohit = [620, 780, 890, 940, 1010, 1095, 1170, 1240, 1160, 1080]

sehwag = [780, 690, 610, 550, 480, 430, 390, 350, 320, 290]

plt.plot(years, kohli, color = "red", linestyle = "--", linewidth = 3, label = "Kohli")
plt.plot(years, rohit, color = "purple", linestyle = ":", linewidth = 3, label = "Rhit")
plt.plot(years, sehwag, color = "yellow", linestyle = "-.", linewidth = 3, label = "Sehwag")
plt.title("Overall runs scored")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.grid(True)
plt.tight_layout()
plt.style.use("ggplot")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as pl
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years)) # index positions
width = 0.25

plt.bar(x - width, sachin, width = width, label = "Sachin")
plt.bar(x, sehwag, width = width, label = "Sehwagh")
plt.bar(x + width, kohli, width = width, label= "Kohli")

plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Runs Scored")
plt.xticks(x, years)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs, color = "#fb2b79")

for i in range(len(players)):
    plt.text(i, runs[i]+30, str(runs[i]), ha= 'center')

plt.xlabel("players")
plt.ylabel("Runs")
plt.title("Comparison")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt 
labels = ["Sachin","Sehwwagh","Kohli","Rohit"]
runs = [1234,2345,4567,3456]
colors = ["#ee0000","#0079f2","#39d039","#ee8925"]
plt.title("Carrer runs of indian players")
plt.pie(runs, labels = labels, colors= colors, wedgeprops= {"edgecolor": 'black', 'linewidth': 2, 'linestyle': '--'})
plt.show()
```

```python
explode = [0.3,0,0,0]
plt.pie(runs, labels = labels, colors=colors, explode=explode, shadow= True, startangle= 140)
plt.title("Exploded result of runs")
plt.show()
```

```python
plt.pie(
 runs,
 labels=labels,
 autopct='%1.1f%%',
 startangle=140
)
plt.title("Career Run Share")
plt.show()
```

```python
import matplotlib.pyplot as plt
activities = ['Studying','Playing','Watching TV', 'Sleeping']
time_spent = [3,2,25]
colors = ['orange','skublue','red','gold']
plt.figure(figsize=(6,6))
plt.pie(time_spent, labels=activities, colors=colors, autopct='%1.1f%%',
startangle=90)
plt.title("After School Activities")
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
